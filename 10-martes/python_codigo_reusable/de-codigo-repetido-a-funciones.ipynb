# Python: De repetición de código a reuso en funciones

Como ejemplo, usemos el cálculo de los componentes de la corriente en la dirección **Este**, `u`, y **Norte**, `v`, tomando la `rapidez` (magnitud del vector de velocidad) y el `angulo` (ángulo en la convención geográfica, en grados, donde los valores aumentan en el sentido del reloj de 0, Norte, a 360):

```
u = rapidez x sin(angulo)
v = rapidez x cos(angulo)
```

## Cálculos individuales, repitiendo la fórmula

````python
import math
````

````python
# En km/h y grados
rapidez1, angulo1 = 1.0, 30.0
````

````python
u = rapidez1 * math.sin(math.radians(angulo1))
v = rapidez1 * math.cos(math.radians(angulo1))

u, v
````

Ahora hacer la transformación de otro vector de velocidad

````python
rapidez2, angulo2 = 2.0, 90.0
````

Tenemos que repetir las fórmulas ...

````python
u = rapidez2 * math.sin(math.radians(angulo2))
v = rapidez2 * math.cos(math.radians(angulo2))

u, v
````

Y si tenemos **otro** vector, a repetir las fórmulas una tercera vez!

**También podríamos cambiar las fórmulas ("vectorizarlas") para que acepten arreglos ("arrays") de `numpy`, pero no lo haremos aquí!**

## Crear una función para ejecutar la conversión

````python
def rapidez_angulo_a_u_v(rapidez, angulo):
    """
    Convertir un vector de velocidad de (rapidez, angulo) a (u, v)
    """
    u = rapidez * math.sin(math.radians(angulo))
    v = rapidez * math.cos(math.radians(angulo))
    return u, v
````

````python
u, v = rapidez_angulo_a_u_v(rapidez1, angulo1)
u, v
````

````python
u, v = rapidez_angulo_a_u_v(rapidez2, angulo2)
u, v
````

## Y si queremos usar la función desde otro notebook?? Módulos!

Podemos crear un módulo ("module") de Python, un archivo separado que contiene la función. Y luego podemos *importar* este módulo aquí, y desde cualquier notebook u otro módulo

````python
import conversion
````

````python
u, v = conversion.rapidez_angulo_a_u_v(rapidez1, angulo1)

u, v
````

Eso funciona bien si todo el código está en el mismo directorio, o si le añadimos a las rutas ("paths") que busca Python, la ruta dónde está este módulo. Usamos `sys.path.append`:

```python
import sys
sys.path.append("/ruta/de/mi/modulo")
```

Esto no es bien portátil.
